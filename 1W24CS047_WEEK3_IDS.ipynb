```python
def print_board(state):
    """Helper function to print the 3x3 puzzle board."""
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()


def dls_8_puzzle(current_state, goal_state, depth, path):
    # Goal check
    if current_state == goal_state:
        return path

    # Depth limit reached
    if depth == 0:
        return None

    idx = current_state.index(0)
    row, col = divmod(idx, 3)

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        nr, nc = row + dr, col + dc

        # Check valid position
        if 0 <= nr < 3 and 0 <= nc < 3:

            new_idx = nr * 3 + nc

            # Create new state
            temp_list = list(current_state)

            temp_list[idx], temp_list[new_idx] = \
                temp_list[new_idx], temp_list[idx]

            next_state = tuple(temp_list)

            # Avoid states already present in current path
            if next_state not in path:

                result = dls_8_puzzle(
                    next_state,
                    goal_state,
                    depth - 1,
                    path + [next_state]
                )

                if result is not None:
                    return result

    return None


def ids_8_puzzle(initial_state,
                  goal_state=(1, 2, 3, 4, 5, 6, 7, 8, 0),
                  max_depth=20):

    # Start from depth 0
    depth = 0

    while depth <= max_depth:

        print("Searching at depth:", depth)

        path = [initial_state]

        result = dls_8_puzzle(
            initial_state,
            goal_state,
            depth,
            path
        )

        # Solution found
        if result is not None:
            return result

        # Increase depth
        depth += 1

    return None


if __name__ == "__main__":

    start_state = (1, 2, 3, 4, 0, 6, 7, 5, 8)

    print("Initial Board:")
    print_board(start_state)

    solution_path = ids_8_puzzle(
        start_state,
        max_depth=8
    )

    if solution_path:

        print(
            f"Solution found in "
            f"{len(solution_path) - 1} steps!\n"
        )

        for step_num, state in enumerate(solution_path):

            print(f"Step {step_num}:")
            print_board(state)

    else:

        print(
            "No solution found within "
            "the specified depth limit."
        )
```
