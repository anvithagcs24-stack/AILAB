```python
print("Anvitha G")
print("1WN24CS047")

from itertools import count

def unify(x, y, subst):
    if x == y:
        return subst
    if x[0].islower() and x not in subst:
        subst[x] = y
        return subst
    if y[0].islower() and y not in subst:
        subst[y] = x
        return subst
    return None

def apply_substitution(literal, subst):
    neg, pred, args = literal
    for _ in range(len(subst) + 1):
        args = tuple(subst.get(a, a) for a in args)
    return (neg, pred, args)

def unify_literals(l1, l2):
    if l1[0] == l2[0] or l1[1] != l2[1]:
        return None

    subst = {}
    for a, b in zip(l1[2], l2[2]):
        subst = unify(a, b, subst)
        if subst is None:
            return None

    return subst

def standardize(clause, suffix):
    variables = set()
    for neg, pred, args in clause:
        for arg in args:
            if arg[0].islower():
                variables.add(arg)

    mapping = {v: v + str(suffix) for v in variables}

    return [
        (neg, pred, tuple(mapping.get(a, a) for a in args))
        for neg, pred, args in clause
    ]

def resolve(c1, c2, step):
    c1 = standardize(c1, step * 2)
    c2 = standardize(c2, step * 2 + 1)

    results = []

    for i, l1 in enumerate(c1):
        for j, l2 in enumerate(c2):
            subst = unify_literals(l1, l2)

            if subst is not None:
                resolvent = []

                for k, lit in enumerate(c1):
                    if k != i:
                        resolvent.append(apply_substitution(lit, subst))

                for k, lit in enumerate(c2):
                    if k != j:
                        resolvent.append(apply_substitution(lit, subst))

                resolvent = list(dict.fromkeys(resolvent))
                results.append((resolvent, subst, i, j))

    return results

def format_literal(lit):
    neg, pred, args = lit
    sign = "¬" if neg else ""
    return sign + pred + "(" + ", ".join(args) + ")"

def format_clause(clause):
    if not clause:
        return "□"
    return " ∨ ".join(format_literal(lit) for lit in clause)

kb = [
    [(True, "food", ("x",)), (False, "likes", ("John", "x"))],

    [(False, "food", ("Apple",))],

    [(False, "food", ("vegetables",))],

    [(True, "eats", ("x", "y")),
     (False, "killed", ("x",)),
     (False, "food", ("y",))],

    [(False, "eats", ("Anil", "Peanuts"))],

    [(False, "alive", ("Anil",))],

    [(True, "alive", ("x",)),
     (True, "killed", ("x",))],

    [(False, "killed", ("x",)),
     (False, "alive", ("x",))]
]

negated_query = [
    (True, "likes", ("John", "Peanuts"))
]

kb.append(negated_query)

print("=" * 65)
print("       FIRST ORDER LOGIC - RESOLUTION")
print("=" * 65)

print("\nINITIAL KNOWLEDGE BASE\n")

for i, clause in enumerate(kb, 1):
    print(f"C{i}: {format_clause(clause)}")

print("\n" + "=" * 65)
print("             RESOLUTION PROOF")
print("=" * 65)

clauses = list(kb)
derived = set()
step = 1
found = False

while step <= 100:
    new_clauses = []

    for i in range(len(clauses)):
        for j in range(i + 1, len(clauses)):
            results = resolve(clauses[i], clauses[j], step)

            for resolvent, subst, _, _ in results:
                key = tuple(sorted(resolvent))

                if key in derived:
                    continue

                derived.add(key)

                print(f"\nStep {step}")
                print(f"Resolving: {format_clause(clauses[i])}")
                print(f"With:      {format_clause(clauses[j])}")

                if subst:
                    print("Unifier:   " + ", ".join(
                        f"{k} = {v}" for k, v in subst.items()
                    ))

                print(f"Result:    {format_clause(resolvent)}")

                if not resolvent:
                    found = True
                    break

                new_clauses.append(resolvent)
                step += 1

            if found:
                break
        if found:
            break

    if found or not new_clauses:
        break

    clauses.extend(new_clauses)

print("\n" + "=" * 65)

if found:
    print("EMPTY CLAUSE (□) DERIVED")
    print("CONTRADICTION FOUND")
    print("\nCONCLUSION: John likes peanuts.")
    print("QUERY PROVED SUCCESSFULLY!")
else:
    print("Resolution proof could not be completed.")

print("=" * 65)
```
