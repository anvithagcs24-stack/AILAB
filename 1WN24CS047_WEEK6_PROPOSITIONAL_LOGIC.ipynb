```python
print("1WN24CS047")
print("Anvitha G")

import itertools

print("P\tQ\tR\tQ->P\tP->~Q\tQVR\tKB")

models = []

for P, Q, R in itertools.product([True, False], repeat=3):

    s1 = (not Q) or P       # Q -> P
    s2 = (not P) or (not Q) # P -> ~Q
    s3 = Q or R             # Q V R

    KB = s1 and s2 and s3

    print(int(P), "\t", int(Q), "\t", int(R), "\t",
          int(s1), "\t", int(s2), "\t",
          int(s3), "\t", int(KB))

    if KB:
        models.append((P, Q, R))

# Check entailments
entails_R = all(R for P, Q, R in models)

entails_R_implies_P = all((not R) or P for P, Q, R in models)

entails_Q_implies_R = all((not Q) or R for P, Q, R in models)

print("\nModels of KB:", models)

print("ii) KB entails R:", entails_R)
print("iii) KB entails R -> P:", entails_R_implies_P)
print("iv) KB entails Q -> R:", entails_Q_implies_R)
```
